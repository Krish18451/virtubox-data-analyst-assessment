```python
Q1 = df_raw["Quantity"].quantile(0.25)
Q3 = df_raw["Quantity"].quantile(0.75)
IQR = Q3 - Q1
upper_limit_qty = Q3 + 1.5 * IQR

print("Quantity upper limit:", upper_limit_qty)
print("Potential quantity outliers:", (df_raw["Quantity"] > upper_limit_qty).sum())
```

```python
# FINAL PROCESSED DATA

df = df_raw.copy()

# 1. Standardize column names
df = df.rename(columns={
    "Invoice": "InvoiceNo",
    "Customer ID": "CustomerID",
    "Price": "UnitPrice"
})

# 2. Correct data types
df["InvoiceNo"] = df["InvoiceNo"].astype(str).str.strip()
df["StockCode"] = df["StockCode"].astype(str).str.strip()
df["Description"] = df["Description"].astype("string").str.strip()
df["Country"] = df["Country"].astype("string").str.strip()

df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
df["CustomerID"] = pd.to_numeric(df["CustomerID"], errors="coerce")

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    dayfirst=True,
    errors="coerce"
)

# 3. Remove exact duplicate rows
before_duplicates = len(df)
df = df.drop_duplicates().copy()
duplicates_removed = before_duplicates - len(df)

# 4. Identify cancelled transactions
df["IsCancelled"] = (
    df["InvoiceNo"].str.startswith("C") |
    (df["Quantity"] < 0)
)

# 5. Create Revenue
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

# 6. Handle missing Description
df["Description"] = df["Description"].fillna("Unknown")

# 7. Create analysis-ready sales dataset
processed = df[
    (~df["IsCancelled"]) &
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)
].copy()

# 8. Keep only useful columns for Processed Data
processed = processed[
    [
        "InvoiceNo",
        "StockCode",
        "Description",
        "Quantity",
        "InvoiceDate",
        "UnitPrice",
        "CustomerID",
        "Country",
        "Revenue"
    ]
]

# 9. Outlier identification using IQR
Q1_qty = processed["Quantity"].quantile(0.25)
Q3_qty = processed["Quantity"].quantile(0.75)
IQR_qty = Q3_qty - Q1_qty
qty_upper = Q3_qty + 1.5 * IQR_qty

Q1_price = processed["UnitPrice"].quantile(0.25)
Q3_price = processed["UnitPrice"].quantile(0.75)
IQR_price = Q3_price - Q1_price
price_upper = Q3_price + 1.5 * IQR_price

qty_outliers = (processed["Quantity"] > qty_upper).sum()
price_outliers = (processed["UnitPrice"] > price_upper).sum()

print("========== PROCESSING SUMMARY ==========")
print("Original rows:", len(df_raw))
print("Duplicates removed:", duplicates_removed)
print("Cleaned rows:", len(df))
print("Processed sales rows:", len(processed))
print("Processed columns:", len(processed.columns))

print("\n========== DATE RANGE ==========")
print("Start:", processed["InvoiceDate"].min())
print("End:", processed["InvoiceDate"].max())

print("\n========== OUTLIERS ==========")
print("Quantity upper limit:", qty_upper)
print("Quantity outliers:", qty_outliers)
print("Price upper limit:", price_upper)
print("Price outliers:", price_outliers)

print("\n========== REVENUE ==========")
print("Total Revenue:", round(processed["Revenue"].sum(), 2))
```
